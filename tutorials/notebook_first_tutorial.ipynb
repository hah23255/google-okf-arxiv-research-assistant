# Notebook-first Tutorial: Pure-OKF ArXiv Assistant

This notebook walks through produce -> validate -> query -> API checks.

## 1) Install dependencies
```bash
uv sync --extra dev
```

````python
from pathlib import Path
import json

data = {
    "paper_id": "2106.09685",
    "title": "LoRA: Low-Rank Adaptation of Large Language Models",
    "abstract": "LoRA adapts large models by injecting trainable low-rank matrices.",
    "url": "https://arxiv.org/abs/2106.09685",
    "categories": ["cs.CL", "cs.LG"],
}
Path('papers.jsonl').write_text(json.dumps(data) + '\n', encoding='utf-8')
print('papers.jsonl created')
````

## 2) Produce bundle
```bash
uv run okf-assistant produce-jsonl --input papers.jsonl --output okf
```

## 3) Validate bundle
```bash
uv run okf-assistant validate --bundle okf
```

## 4) Query bundle
```bash
uv run okf-assistant query --bundle okf --question "What is LoRA?"
```

## 5) API check
```bash
uv run uvicorn google_okf_arxiv_assistant.api:app --reload
curl -s -X POST http://127.0.0.1:8000/query -H 'Content-Type: application/json' -d '{"query":"What is LoRA?","top_k":5}'
```

## 6) Launch Streamlit frontend
```bash
uv run streamlit run app.py
```
Use the sidebar to set API Base URL and top_k.
